```python
df = sparcs_combined.copy()

# Drop columns
df.drop(columns=[
    'Discharge Year',
    'Payment Typology 2',
    'Payment Typology 3',
    'Total Charges',
    'Patient Disposition',
    'Facility Name',
    'CCSR Diagnosis Description',
    'CCSR Procedure Description',
    'APR DRG Description',
    'APR MDC Description',
    'APR Severity of Illness Description',
], inplace=True)

# Rename columns
rename_map = {
    'Health Service Area': 'health_service_area',
    'Hospital County': 'hospital_county',
    'Permanent Facility Id': 'facility_id',
    'Age Group': 'age_group',
    'Zip Code': 'zip_code',
    'Gender': 'gender',
    'Race': 'race',
    'Ethnicity': 'ethnicity',
    'Length of Stay': 'length_of_stay',
    'Type of Admission': 'admission_type',
    'CCSR Diagnosis Code': 'ccsr_dx_code',
    'CCSR Procedure Code': 'ccsr_px_code',
    'APR MDC Code': 'apr_mdc_code',
    'APR DRG Code': 'apr_drg_code',
    'APR Severity of Illness Code': 'apr_severity_code',
    'APR Medical Surgical Description': 'apr_med_surg_desc',
    'APR Risk of Mortality': 'apr_mortality_risk',
    'Total Charges': 'total_charges',
    'Medicaid': 'payer_medicaid',
    'Medicare': 'payer_medicare',
    'Private Health Insurance': 'payer_private_insurance',
    'Self-Pay': 'payer_self_pay',
    'Blue Cross/Blue Shield': 'payer_blue_cross',
    'Miscellaneous/Other': 'payer_other',
    'Federal/State/Local/VA': 'payer_gov_va',
    'Department of Corrections': 'payer_corrections',
    'Managed Care, Unspecified': 'payer_managed_care',
    'Number of Payment Typologies': 'num_payment_types',
    'Emergency Department Indicator': 'emergency_dept_indicator'
}

df.rename(columns=rename_map, inplace=True)
```

```python
df.info()
```

```python
# Remove all NaN/"missing"/"U"/"not available"/"unknown" values
# Define missing_values

df_clean = df.copy()

# Define all missing value patterns (lowercase for comparison)
missing_values = [
    'nan', 'none', 'null', 'missing', 'unknown', 'u',
    'not available', 'n/a', 'na', '', ' ', 'undefined',
    'no data', 'not provided', '--', '?', 'none available'
]

# For each column, check if values match missing patterns (case-insensitive)
for col in df_clean.columns:
    if df_clean[col].dtype == 'object':
        # Create a mask for missing values (case-insensitive comparison)
        mask = df_clean[col].astype(str).str.lower().str.strip().isin(missing_values)
        # Replace with actual NaN, preserving other values' capitalization
        df_clean.loc[mask, col] = np.nan

# Drop rows with any NaN values (catches both replaced strings and existing NaNs)
df_clean = df_clean.dropna()

print(f"Original shape: {df.shape}")
print(f"Cleaned shape: {df_clean.shape}")
print(f"Rows removed: {df.shape[0] - df_clean.shape[0]}")
```

```python
# Convert LOS dtype to string for standardizing reporting
df_clean['length_of_stay'] = df_clean['length_of_stay'].astype('string')

# Convert any 120+ LOS values to 120 for ML
df_clean['length_of_stay'] = df_clean['length_of_stay'].apply(lambda x: '120' if str(x).startswith('120') else x)

# Convert LOS back to integer
df_clean['length_of_stay'] = df_clean['length_of_stay'].astype('int')

print(df_clean['length_of_stay'].value_counts())
```

```python
# Check unique in specific cols
# Check each column
for col in df_clean.columns:
    unique_count = df_clean[col].nunique()

    if unique_count < 30:
        print(f"\n{col} ({unique_count} unique values):")
        print(df_clean[col].unique())
        print("-" * 50)
```

```python
# drop all unknown entries from ethnicity column
df_clean = df_clean[df_clean['ethnicity'] != 'Unknown']
df_clean = df_clean[df_clean['apr_mortality_risk'] != 'Undetermined']
```

```python
# convert dtypes
df_clean = df_clean.copy()
df_clean['emergency_dept_indicator'] = (
    df_clean['emergency_dept_indicator'].astype('bool')
)

string_cols = [
    'race', 'ethnicity', 'apr_mortality_risk', 'apr_med_surg_desc',
    'Payment Typology 1'
]

df_clean[string_cols] = df_clean[string_cols].astype('object')
df_clean['facility_id'] = df_clean['facility_id'].astype('Int64').astype('object')
df_clean['apr_drg_code'] = df_clean['apr_drg_code'].astype('object')
df_clean['apr_mdc_code'] = df_clean['apr_mdc_code'].astype('object')
df_clean['apr_severity_code'] = df_clean['apr_severity_code'].astype('object')

print(df_clean.dtypes)
```

```python
df_clean = pd.read_feather('/content/drive/MyDrive/Sparcs_Datafiles/preprocessing_v2/sparcs_clean_v2.feather')
```

```python
# Encode variables
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
import pandas as pd

# Create a copy to work with
df_encoded = df_clean.copy()

# 1. LABEL ENCODING
# Features that should use label encoding
label_encode_cols = [
    'gender',
    'apr_med_surg_desc',
    'emergency_dept_indicator',
    'hospital_county',
    'facility_id',
    'zip_code',
    'ccsr_dx_code',
    'ccsr_px_code',
    'apr_drg_code',
    'apr_mdc_code',
    'admission_type'
]

# Apply label encoding
le = LabelEncoder()
for col in label_encode_cols:
    if col in df_encoded.columns:
        df_encoded[col] = le.fit_transform(df_encoded[col].astype(str))

# Handle emergency_dept_indicator (boolean to 0/1)
if 'emergency_dept_indicator' in df_encoded.columns:
    df_encoded['emergency_dept_indicator'] = df_encoded['emergency_dept_indicator'].astype(int)

# 2. ONE-HOT ENCODING
# Features that should use one-hot encoding
onehot_encode_cols = [
    'health_service_area',
    'ethnicity',
    'race',
    'apr_severity_code',
    'Payment Typology 1'
]

# Apply one-hot encoding
df_encoded = pd.get_dummies(df_encoded, columns=onehot_encode_cols, prefix=onehot_encode_cols, drop_first=False)

# 3. ORDINAL ENCODING for APR Risk of Mortality
# Sort values and encode from 0 to N-1
if 'apr_mortality_risk' in df_encoded.columns:
    # Define the order
    risk_order = ['Minor', 'Moderate', 'Major', 'Extreme']
    risk_mapping = {risk: idx for idx, risk in enumerate(risk_order)}
    df_encoded['apr_mortality_risk'] = df_encoded['apr_mortality_risk'].map(risk_mapping)

# 4. ORDINAL ENCODING for Age Group
if 'age_group' in df_encoded.columns:
    age_order = ['0-17', '18-29', '30-49', '50-69', '70 or Older']
    age_mapping = {age: idx for idx, age in enumerate(age_order)}
    df_encoded['age_group'] = df_encoded['age_group'].map(age_mapping)

# Check the results
print("\nEncoded DataFrame shape:", df_encoded.shape)
print("\nEncoded DataFrame columns:", df_encoded.columns.tolist())
print("\nData types:\n", df_encoded.dtypes)
```
